# A written rule versus a learned one

This notebook goes with the Detailed lesson **Artificial Intelligence: Ideas, Approaches and History**, section 5.

We flag spam emails using one clue: the number of suspicious phrases in each message. First we try a rule a programmer guessed, then we let the data choose the rule, then we test both on emails neither has seen.

Run each cell in order with **Shift + Enter**.

## 1. The labelled emails

Each email is (number of suspicious phrases, label). The label is 1 for spam and 0 for not spam.

```python
training = [
    (3, 1), (4, 1), (5, 1), (6, 1), (3, 1),   # spam
    (0, 0), (1, 0), (1, 0), (2, 0), (0, 0),   # not spam
]
test = [(3, 1), (2, 1), (2, 0), (0, 0)]       # new emails, kept aside

print(len(training), "training emails,", len(test), "test emails")
```

## 2. The rule-based way

A programmer guesses: flag an email if it has 2 or more suspicious phrases.

```python
def flag(phrases, cut_off):
    return 1 if phrases >= cut_off else 0

def mistakes(emails, cut_off):
    return sum(1 for phrases, label in emails if flag(phrases, cut_off) != label)

print("Guessed rule (cut-off 2):", mistakes(training, 2), "mistake(s) on the training emails")
```

## 3. The learning way

Try every possible cut-off and count the mistakes on the training emails. Keep the cut-off with the fewest.

```python
for cut_off in range(0, 8):
    print("flag at", cut_off, "or more:", mistakes(training, cut_off), "mistake(s)")

learned = min(range(0, 8), key=lambda c: mistakes(training, c))
print()
print("Learned cut-off:", learned)
```

## 4. Checking on new emails

A rule that fits the training emails can still fail on new ones. Accuracy is the share of test emails it gets right.

```python
def accuracy(emails, cut_off):
    right = sum(1 for phrases, label in emails if flag(phrases, cut_off) == label)
    return right / len(emails)

for name, cut_off in [("guessed", 2), ("learned", learned)]:
    print(f"{name} rule, cut-off {cut_off}: test accuracy {accuracy(test, cut_off):.0%}")

for phrases, label in test:
    print(f"  {phrases} phrases, really {'spam' if label else 'not spam'}, learned rule says {'spam' if flag(phrases, learned) else 'not spam'}")
```
